```python
from diffusers import StableDiffusionPipeline
import torch

model_id = "sd-legacy/stable-diffusion-v1-5"

pipe = StableDiffusionPipeline.from_pretrained(
    model_id,
    torch_dtype=torch.float16
)

pipe = pipe.to("cuda")


#Prompt 1
# prompt = """
# giant croissant floating,
# a quiet vintage town,
# surreal vintage paper collage,
# dreamlike composition,
# 1950s magazine cutout style,
# soft muted colors,
# visible paper texture
# """

#Prompt 2

# prompt = """
# giant croissant floating,
# a calm ocean,
# surreal vintage paper collage,
# dreamlike composition,
# 1950s magazine cutout style,
# soft muted colors,
# visible paper texture
# """

#Prompt 3

prompt = """
giant croissant floating,
a quiet grassy field with tiny people having a picnic,
surreal vintage paper collage,
dreamlike composition,
1950s magazine cutout style,
soft muted colors,
visible paper texture
"""



negative_prompt = """
text, watermark, blurry, modern, 3D render
"""

image = pipe(
    prompt=prompt,
    negative_prompt=negative_prompt,
    guidance_scale=9,
    num_inference_steps=40
).images[0]

image.save("collage_1.png")
```
